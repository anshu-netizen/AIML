Q1. What does the shape of X represent?

 The shape of X represents the number of samples and features in the dataset. Each row corresponds to one image, and each column represents a pixel value.

Q2. What does the shape of y represent?

The shape of y represents the number of labels corresponding to each input sample. Each value in y is the class label (0 or 1) for each image.

Q3. What does the weights array represent?

The weights array represents the importance of each feature. Each weight determines how strongly a pixel influences the prediction of whether the digit if 0 or 1.

Q4. Why initialize weights to zero?
Weights are initialized to zero to provide a neutral starting point for learning. This allows the model to adjust weights gradually based on errors. However, starting with zero may slow convergence since all features initially contribute equally.

Q5. Purpose of np.dot(X[i], weighst)+bias

This computes the weighted sum of inputs plus bias. It represents the activation value of the perceptron before applying the step function.

Q6. What happens when prediction is wrong?

When the prediction is incorrect, the weights are updated using w = w + η(y − ŷ)X
and the bias is updated using b = b + η(y − ŷ),
this shifts the decicion boundary to reduce future errors.

Q7. Why is final accuracy important?

Final accuracy measures how well the model classifies the dataset. It indicates the effectiveness of the learned decision boundary. For 0 vs 1 classification, high accuracy is expected because the data is mostly linearly separable.

Q8. What does missclassified_idx store?

It stores the indices of all incorrectly classified samples. These indices are used to visualize the misclassified images.

Q9. Interpretation of “All images were correctly classified!”

This means the model perfectly classified all samples.
It indicates that the dataset is linearly separable and the perceptron successfully learned the decision boundary.
