```python
from PIL import Image
image=Image.open("/content/drive/MyDrive/AIML/Data/lena_gray.gif")
image_colored=Image.open("/content/drive/MyDrive/AIML/Data/ok.jpeg")
display(image)
```

```python
print ("Image Format: ", image.format)
print ("Image Size: ", image.size)
print ("Image Mode: ", image.mode)
```

```python
width , height = image.size
channels = len( image.getbands () )
print ( f" Image shape (RGB): ({ height } , { width } , { channels })")
image_size_colored = width * height *3
print ( f" Image size (RGB): { image_size_colored }")
```

```python
import numpy as np
import matplotlib.pyplot as plt

image_array_colored = np.array(image_colored)
print("Shape of the image array:", image_array_colored.shape)

# red channel
red_channel = image_array_colored[:, :, 0]
print(red_channel)
plt.imshow(red_channel, cmap='gray')
plt.title("Red Channel")
plt.axis('off')
plt.show()

# green channel
green_channel = image_array_colored[:, :, 1]   # FIX: index is 1
print(green_channel)
plt.imshow(green_channel, cmap='gray')
plt.title("Green Channel")
plt.axis('off')
plt.show()

# blue channel
blue_channel = image_array_colored[:, :, 2]
print(blue_channel)
plt.imshow(blue_channel, cmap='gray')
plt.title("Blue Channel")
plt.axis('off')
plt.show()
```
